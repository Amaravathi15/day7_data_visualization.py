```python
##Subplots (multiple charts in one figure)
##Four Subplots (2×2)

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

fig, ax = plt.subplots(2, 2, figsize=(8, 6))

# Plot 1
ax[0, 0].plot(x, [1, 2, 3, 4, 5])
ax[0, 0].set_title("Line Plot")

# Plot 2
ax[0, 1].bar(x, [5, 4, 3, 2, 1])
ax[0, 1].set_title("Bar Chart")

# Plot 3
ax[1, 0].scatter(x, [2, 4, 6, 8, 10])
ax[1, 0].set_title("Scatter Plot")

# Plot 4
ax[1, 1].pie([20, 30, 50], labels=["A", "B", "C"], autopct="%1.1f%%")
ax[1, 1].set_title("Pie Chart")

plt.tight_layout()
plt.show()
```

```python
1.## Line/bar chart showing Sales by Category

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("RetailSales Day7.csv")

# Total sales by category
sales_by_category = df.groupby("Category")["Sales"].sum()

#plot
plt.figure(figsize=(8,6))

sales_by_category.plot(kind="bar",color="orange")
plt.title("sales by category")
plt.xlabel("category")
plt.ylabel("total sales")
plt.xticks(rotation=45)
plt.grid(axis="y",linestyle="--",alpha=0.7)
plt.show()
print("Insight:")
print("This chart compares total sales across different product categories.")
```

```python
##5 Boxplot to detect outliers in Sales or Profit

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#load Data
df = pd.read_csv("RetailSales Day7.csv")

#Box plot for sales

plt.figure(figsize=(6,4))

sns.boxplot(
    y=df["Sales"],
    color="skyblue"
)

plt.title("Boxplot of Sales")
plt.ylabel("Sales")

plt.show()

#Box plot for Profit

plt.figure(figsize=(6,4))

sns.boxplot(
    y=df["Profit"],
    color="skyblue"
)

plt.title("Boxplot of Profit")
plt.ylabel("Sales")

plt.show()
print("Insight:")
print("The box represents the middle 50% of sales values\
The line inside the box is the median\
Any dots outside the whiskers are outliers.")
```

```python
##detect outliers in Sales
# Calculate Q1, Q3, and IQR

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#load Data
df = pd.read_csv("RetailSales Day7.csv")

## Calculate the first quartile (25%) and third quartile (75%)

Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)

# Calculate the Interquartile Range (IQR)
IQR = Q3 - Q1

# Define lower and upper limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Find outliers
outliers = df[(df["Sales"] < lower_limit) | (df["Sales"] > upper_limit)]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of Outliers =", len(outliers))
print("\nOutliers in Sales:")
print(outliers)
# Display results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of Outliers:", len(outliers))

print("\nOutliers in Sales:")
print(outliers)
print("Insight:")
print("The boxplot helps identify outliers in sales values.")
```

```python
6.## Seaborn correlation heatmap of the numeric columns
##Include a correlation heatmap (Sales, Quantity, Discount, Profit)


import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("RetailSales Day7.csv")

# Select only numeric columns
numeric_df = df.select_dtypes(include="number")

# Correlation matrix
corr = numeric_df.corr()
print(corr)

# Plot heatmap
plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
plt.title("Retail Sales Correlation Heatmap")
plt.show()
print("Insight:")
print("The heatmap displays correlations between Sales, Quantity, Discount, and Profit.")
```

```python
7.## subplots to show 2+ related charts in one figure
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("RetailSales Day7.csv")

# Group data by Category
sales_by_category = df.groupby("Category")["Sales"].sum()
profit_by_category = df.groupby("Category")["Profit"].sum()

# Create a 2x2 subplot layout
plt.figure(figsize=(12, 8))

# 1. Bar Chart - Sales by Category
plt.subplot(2, 2, 1)
plt.bar(sales_by_category.index, sales_by_category.values, color="skyblue")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=30)

# 2. Line Chart - Profit by Category
plt.subplot(2, 2, 2)
plt.plot(profit_by_category.index, profit_by_category.values,
         marker="o", color="green")
plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.xticks(rotation=30)

# 3. Histogram - Sales Distribution
plt.subplot(2, 2, 3)
plt.hist(df["Sales"], bins=10, color="orange", edgecolor="black")
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")


# 4. Scatter Plot - Sales vs Profit
plt.subplot(2, 2, 4)
plt.scatter(df["Sales"], df["Profit"], color="red", alpha=0.6)
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

# Adjust spacing
plt.tight_layout()

# Display all charts
plt.show()

print("Insight:")
print("The subplot compares category-wise sales and  profit in one figure.")
```

```python
# Pie Chart (Category Sales)

category_sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(7,7))

plt.pie(category_sales,
        labels=category_sales.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Sales Share by Category")

plt.savefig("pie_category_sales.png")

plt.show()

print("Insight:")
print("The pie chart shows each category's percentage contribution to total sales.")

```
